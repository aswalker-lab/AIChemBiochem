# PCA and Unsupervised Learning
The PCA we discussed last time can also be used for unsupervised learning, clusters can sometimes be seen in the reduced data. We will look at this for reduced representation of molecular features

```python
import pandas as pd

#load and look at data
df = pd.read_csv("pubchem_unsupervised_example.csv")
#sometimes data will have incorrect values, if they are supposed be a number but not they will often be annotated with NaN for
#not a number we can use dropna() to remove rows with NaN values
df = df.dropna()
df
```

```python
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

#assemble the pipeline
pipe = Pipeline([
    ("scaler", StandardScaler()),   
    ("pca", PCA(n_components=2))
])

#note this dataset already has some descriptors so we do not need to calculate them ourselves with rdkit
reduced_data = pipe.fit_transform(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])

plt.figure(figsize=(8, 6))
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.scatter(reduced_data[:,0],reduced_data[:,1])
plt.show()
```

# tSNE and UMAP
Compare the tSNE and UMAP results to PCA - which has more obvious clusters?
<br/>
Note that these methods have hyperparameters that will affect the result
<br/>
tSNE and UMAP can also be applied to fingerprints

```python
#tSNE and UMAP applied to descriptors
from sklearn.manifold import TSNE
import umap

X_tsne = TSNE(n_components=2, learning_rate='auto',
                  init='random', perplexity=3).fit_transform(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])
plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:,0],X_tsne[:,1])
plt.title("tSNE on molecular descriptors")
plt.show()

X_umap = umap.UMAP(n_neighbors=5, random_state=42).fit(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])
plt.figure(figsize=(8, 6))
plt.scatter(X_umap.embedding_[:,0],X_umap.embedding_[:,1])
plt.title("UMAP  on molecular descriptors")
plt.show()
```

# K-Means Clustering

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=0, n_init="auto").fit(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])

plt.figure(figsize=(8, 6))
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.scatter(reduced_data[:,0],reduced_data[:,1],c=kmeans.labels_, cmap='viridis')
plt.title("PCA on Descriptors, K-Means Clustering")
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:,0],X_tsne[:,1],c=kmeans.labels_, cmap='viridis')
plt.title("tSNE on Descriptors, K-Means Clustering")
plt.show()
```

# Affinity Propagation

```python
from sklearn.cluster import AffinityPropagation

affinity_prop = AffinityPropagation(random_state=5).fit(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])

plt.figure(figsize=(8, 6))
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.scatter(reduced_data[:,0],reduced_data[:,1],c=affinity_prop.labels_, cmap='viridis')
plt.title("PCA on Descriptors, Affinity Propagation Clustering")
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:,0],X_tsne[:,1],c=affinity_prop.labels_, cmap='viridis')
plt.title("tSNE on Descriptors, Affinity Propagation Clustering")
plt.show()
```

# DBSCAN and HDBSCAN

```python
from sklearn.cluster import DBSCAN, HDBSCAN

dbscan_clustering = DBSCAN(eps=3, min_samples=2).fit(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])
hdb_clustering = HDBSCAN(copy=True, min_cluster_size=2).fit(df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]])

plt.figure(figsize=(8, 6))
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.scatter(reduced_data[:,0],reduced_data[:,1],c=dbscan_clustering.labels_, cmap='viridis')
plt.title("PCA on Descriptors, DBSCAN Clustering")
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:,0],X_tsne[:,1],c=dbscan_clustering.labels_, cmap='viridis')
plt.title("tSNE on Descriptors, DBSCAN Clustering")
plt.show()

plt.figure(figsize=(8, 6))
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.scatter(reduced_data[:,0],reduced_data[:,1],c=hdb_clustering.labels_, cmap='viridis')
plt.title("PCA on Descriptors, HDBSCAN Clustering")
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:,0],X_tsne[:,1],c=hdb_clustering.labels_, cmap='viridis')
plt.title("tSNE on Descriptors, HDBSCAN Clustering")
plt.show()
```

# Butina

```python
# adapted from https://zoehlerbz.medium.com/molecule-clustering-using-the-butina-algorithm-application-with-python-and-rdkit-ffb5b5721447
# you can set a distance threshold  for clusters in the Butina clustering algorithm
from rdkit.ML.Cluster import Butina
from rdkit.Chem import Descriptors, AllChem, DataStructs
from rdkit import Chem

def butina(mols):
   matrix = []
   fpgen_morgan = AllChem.GetMorganGenerator(radius=2,fpSize=2048)
   print(mols)
   ind_list = mols.index.tolist()
   for i in ind_list:
       for j in ind_list[0:ind_list.index(i)]:
           fp1 = fpgen_morgan.GetFingerprint(mols[i])
           fp2 = fpgen_morgan.GetFingerprint(mols[j])
           similarity = DataStructs.TanimotoSimilarity(fp1, fp2)
           matrix.append(1 - similarity)

   clusters = Butina.ClusterData(data=matrix, nPts=len(ind_list), distThresh=.4, isDistData=True)
   clusters = sorted(clusters, key=len, reverse=True)

   return clusters

df["mol"] = df["SMILES"].apply(Chem.MolFromSmiles)
clusters = butina(df['mol'])

cluster_list = []
cluster_dic = {}
for indice in range(len(clusters)):
   for registro in clusters[indice]:
       cluster_dic[registro] = indice
for registro in cluster_dic:
    cluster_list.append(cluster_dic[registro])
        
plt.figure(figsize=(8, 6))
plt.xlabel("PC 1")
plt.ylabel("PC 2")
plt.scatter(reduced_data[:,0],reduced_data[:,1],c=cluster_list, cmap='viridis')
plt.title("PCA on Descriptors, Butina Clustering")
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:,0],X_tsne[:,1],c=cluster_list, cmap='viridis')
plt.title("tSNE on Descriptors, Butina Clustering")
plt.show()
```

# Metrics
We will look at the Silhouette score and other metrics calculated on the descriptors. We will also look at the Silhouette score calculated on the Tanimoto similarity converted to a distance.

```python
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
import numpy as np

X = df[["MolecularWeight","XLogP","TPSA","HBondDonorCount","HBondAcceptorCount","RotatableBondCount","HeavyAtomCount","Complexity"]]
print("Silhouette Scores on Descriptors")
print("kmeans " + str(silhouette_score(X, kmeans.labels_)))
print("Affinity Propagation " + str(silhouette_score(X, affinity_prop.labels_)))
print("DBSCAN " + str(silhouette_score(X, dbscan_clustering.labels_)))
print("HDBSCAN " + str(silhouette_score(X, hdb_clustering.labels_)))
print("Butina " + str(silhouette_score(X, cluster_list)))

print("Calinski Harabasz Scores on Descriptors")
print("kmeans " + str(calinski_harabasz_score(X, kmeans.labels_)))
print("Affinity Propagation " + str(calinski_harabasz_score(X, affinity_prop.labels_)))
print("DBSCAN " + str(calinski_harabasz_score(X, dbscan_clustering.labels_)))
print("HDBSCAN " + str(calinski_harabasz_score(X, hdb_clustering.labels_)))
print("Butina " + str(calinski_harabasz_score(X, cluster_list)))

print("Davies Bouldin Score on Descriptors")
print("kmeans " + str(davies_bouldin_score(X, kmeans.labels_)))
print("Affinity Propagation " + str(davies_bouldin_score(X, affinity_prop.labels_)))
print("DBSCAN " + str(davies_bouldin_score(X, dbscan_clustering.labels_)))
print("HDBSCAN " + str(davies_bouldin_score(X, hdb_clustering.labels_)))
print("Butina " + str(davies_bouldin_score(X, cluster_list)))

ind_list = df.index.tolist()
dist_matrix = np.zeros((len(ind_list),len(ind_list)))

fpgen_morgan = AllChem.GetMorganGenerator(radius=2,fpSize=2048)
for i in range(0, len(ind_list)):
    for j in range(0, len(ind_list)):
        fp1 = fpgen_morgan.GetFingerprint(df["mol"][ind_list[i]])
        fp2 = fpgen_morgan.GetFingerprint(df["mol"][ind_list[j]])
        similarity = DataStructs.TanimotoSimilarity(fp1, fp2)
        dist_matrix[i,j] = 1-similarity

print("Silhouette Score on Tanimoto Distance")
print("kmeans " + str(silhouette_score(dist_matrix, kmeans.labels_, metric='precomputed')))
print("Affinity Propagation " + str(silhouette_score(dist_matrix, affinity_prop.labels_, metric='precomputed')))
print("DBSCAN " + str(silhouette_score(dist_matrix, dbscan_clustering.labels_, metric='precomputed')))
print("HDBSCAN " + str(silhouette_score(dist_matrix, hdb_clustering.labels_, metric='precomputed')))
print("Butina " + str(silhouette_score(dist_matrix, cluster_list, metric='precomputed')))
```

# Choosing number of clusters with metrics
For methods like K-means that requires choice of number of clusters, we can see how different cluster sizes effect the Silhouette score, the maximum of the curve will be the ideal number of clusters.

```python
silhouette_scores = []
silhouette_scores_dist = []
for k in range(2,50):
    kmeans = KMeans(n_clusters=k, random_state=0, n_init="auto").fit(X)
    silhouette_scores.append(silhouette_score(X, kmeans.labels_))
    silhouette_scores_dist.append(silhouette_score(dist_matrix, kmeans.labels_, metric='precomputed'))
    
plt.plot(range(2,50), silhouette_scores)
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores Descriptors for Different K")
plt.show()

plt.plot(range(2,50), silhouette_scores_dist)
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores Tanimoto Dist for Different K")
plt.show()
```
