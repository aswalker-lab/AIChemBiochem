# ChemBERTa fine-tuning comparison: aqueous solubility

This notebook finetunes **pretrained ChemBERTa** for predicting Delaney solubility.

**Four approaches:** (1) frozen ChemBERTa backbone; (2) freeze the first half of the encoder blocks; (3) full fine-tuning; (4) LoRA in all encoder blocks. 

**Checkpoint:** [`DeepChem/ChemBERTa-77M-MLM`](https://huggingface.co/DeepChem/ChemBERTa-77M-MLM). "77M" refers to the **pretraining molecule count**, not parameter count. This checkpoint has **three encoder layers**. The partial-freeze condition freezes the **first `floor(3/2)=1` layer** and leaves the other two trainable; embeddings are frozen.

## Load the data

```python
import gc
import time
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import psutil
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel
from peft import LoraConfig, get_peft_model

SEED = 42
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
TARGET = "measured log(solubility:mol/L)"
SMILES_COL = "SMILES"
BATCH_SIZE = 32
EPOCHS = 15          # Same number of epochs for each method
MAX_TOKENS = 256     # Covers most Delaney SMILES; longer sequences are truncated

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(SEED)
df = pd.read_csv("delaney.csv")
df = df.dropna(subset=[SMILES_COL, TARGET]).copy()
train_df, val_df = train_test_split(df, test_size=0.2, random_state=SEED)
print(f"Device: {DEVICE}, training: {len(train_df)}, validation: {len(val_df)}")
print(f"Columns: {list(df.columns)}")
print(f"PyTorch: {torch.__version__}")
```

## ChemBERTa tokenizer and data loaders
Use the pretrained tokenizer rather than the custom SMILES vocabulary. Dynamic padding reduces work on shorter batches.

```python
tokenizer = AutoTokenizer.from_pretrained(CHECKPOINT)
max_supported = tokenizer.model_max_length
if isinstance(max_supported, int) and max_supported < 100000:
    MAX_TOKENS = min(MAX_TOKENS, max_supported)

class SolubilityDataset(Dataset):
    def __init__(self, frame):
        self.smiles = frame[SMILES_COL].astype(str).tolist()
        self.targets = frame[TARGET].astype(float).tolist()
    def __len__(self):
        return len(self.smiles)
    def __getitem__(self, idx):
        return self.smiles[idx], self.targets[idx]

def collate_batch(rows):
    smiles, labels = zip(*rows)
    tokens = tokenizer(list(smiles), padding=True, truncation=True,
                       max_length=MAX_TOKENS, return_tensors="pt")
    tokens["labels"] = torch.tensor(labels, dtype=torch.float32)
    return tokens

# The same train ordering is recreated by resetting the seed for each strategy.
def make_loaders():
    gen = torch.Generator().manual_seed(SEED)
    train_loader = DataLoader(SolubilityDataset(train_df), batch_size=BATCH_SIZE,
                              shuffle=True, generator=gen, collate_fn=collate_batch)
    val_loader = DataLoader(SolubilityDataset(val_df), batch_size=BATCH_SIZE,
                            shuffle=False, collate_fn=collate_batch)
    return train_loader, val_loader

n_truncated = sum(len(tokenizer(s, add_special_tokens=True, truncation=False)["input_ids"]) > MAX_TOKENS
                  for s in df[SMILES_COL].astype(str))
print(f"Maximum tokenized length: {MAX_TOKENS}; truncated molecules: {n_truncated}/{len(df)}")
```

## ChemBERTa regression model and fine-tuning configurations


```python
#class for regression head
class ChemBERTaRegressor(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = AutoModel.from_pretrained(CHECKPOINT, add_pooling_layer=False)
        self.regressor = nn.Sequential(
            nn.Linear(self.encoder.config.hidden_size, 64),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(64, 1),
        )

    def forward(self, input_ids, attention_mask):
        output = self.encoder(input_ids=input_ids, attention_mask=attention_mask)
        hidden = output.last_hidden_state
        # Ignore special tokens as well as padding when mean-pooling.
        valid = attention_mask.bool()
        for token_id in tokenizer.all_special_ids:
            valid = valid & (input_ids != token_id)
        weights = valid.unsqueeze(-1).to(hidden.dtype)
        pooled = (hidden * weights).sum(1) / weights.sum(1).clamp(min=1)
        return self.regressor(pooled).squeeze(-1)

STRATEGIES = ["Frozen backbone", "Freeze first half", "Full fine-tuning", "LoRA all layers"]
LEARNING_RATES = {
    "Frozen backbone": 1e-3,
    "Freeze first half": 2e-4,
    "Full fine-tuning": 2e-4,
    "LoRA all layers": 3e-4,
}

def build_model(strategy):
    model = ChemBERTaRegressor()
    layers = model.encoder.encoder.layer
    n_layers = len(layers)
    if strategy == "Frozen backbone":
        for p in model.encoder.parameters():
            p.requires_grad = False
    elif strategy == "Freeze first half":
        for p in model.encoder.parameters():
            p.requires_grad = False
        # For 3 layers, freeze layer 0; train layers 1 and 2.
        for block in layers[n_layers // 2:]:
            for p in block.parameters():
                p.requires_grad = True
    elif strategy == "Full fine-tuning":
        pass
    elif strategy == "LoRA all layers":
        config = LoraConfig(r=8, lora_alpha=16, lora_dropout=0.05,
                            target_modules=["query", "value"], bias="none")
        model.encoder = get_peft_model(model.encoder, config)
    else:
        raise ValueError(strategy)
    # The new regression head always remains trainable.
    model = model.to(DEVICE)
    return model

def parameter_counts(model):
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total = sum(p.numel() for p in model.parameters())
    return trainable, total

probe = build_model("Freeze first half")
print(f"Encoder blocks: {len(probe.encoder.encoder.layer)}")
for i, layer in enumerate(probe.encoder.encoder.layer):
    print(f"Layer {i}: {sum(p.numel() for p in layer.parameters() if p.requires_grad):,} trainable parameters")
del probe
if DEVICE.type == "cuda":
    torch.cuda.empty_cache()
```

## 5. Train and validate one epoch
No mixed precision, inference tensors, embedding caching, or early stopping. Every approach trains for exactly `EPOCHS` epochs. The validation metrics refer to the same held-out molecules. We record total training and validation time separately for each epoch.

```python
def sync_cuda():
    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

def one_training_epoch(model, loader, optimizer, criterion):
    model.train()
    predictions, targets = [], []
    for batch in loader:
        ids = batch["input_ids"].to(DEVICE)
        mask = batch["attention_mask"].to(DEVICE)
        y = batch["labels"].to(DEVICE)
        optimizer.zero_grad(set_to_none=True)
        predicted = model(ids, mask)
        loss = criterion(predicted, y)
        loss.backward()
        optimizer.step()
        predictions.append(predicted.detach().cpu().numpy())
        targets.append(y.detach().cpu().numpy())
    y_true = np.concatenate(targets)
    y_pred = np.concatenate(predictions)
    return float(np.sqrt(mean_squared_error(y_true, y_pred))), float(r2_score(y_true, y_pred))

@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    predictions, targets = [], []
    for batch in loader:
        predicted = model(batch["input_ids"].to(DEVICE),
                          batch["attention_mask"].to(DEVICE))
        predictions.append(predicted.cpu().numpy())
        targets.append(batch["labels"].numpy())
    y_true = np.concatenate(targets)
    y_pred = np.concatenate(predictions)
    return float(np.sqrt(mean_squared_error(y_true, y_pred))), float(r2_score(y_true, y_pred)), y_true, y_pred
```

## 6. Compare the four strategies
Peak GPU memory is `max_memory_allocated` (not reserved memory); CPU RSS measures process memory and is not the same as GPU memory. Each model is deleted between runs. Training and validation time are synchronized for CUDA. All methods have the **same number of epochs**, rather than the same wall-clock budget; learning rates are configurable and differ to suit the fine-tuning strategies.

```python
history_rows = []
summary_rows = []
prediction_results = {}
process = psutil.Process()

for strategy in STRATEGIES:
    set_seed(SEED)
    gc.collect()
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    model = build_model(strategy)
    train_loader, val_loader = make_loaders()
    optimizer = torch.optim.AdamW(
        (p for p in model.parameters() if p.requires_grad),
        lr=LEARNING_RATES[strategy], weight_decay=0.01)
    criterion = nn.MSELoss()
    trainable, total = parameter_counts(model)
    best_rmse = float("inf")
    best_epoch = None
    best_predictions = None
    train_seconds = 0.0
    val_seconds = 0.0
    peak_cpu_mb = process.memory_info().rss / (1024**2)
    print(f"\n{strategy}: {trainable:,}/{total:,} trainable parameters")

    for epoch in range(1, EPOCHS + 1):
        sync_cuda()
        start = time.perf_counter()
        train_rmse, train_r2 = one_training_epoch(model, train_loader, optimizer, criterion)
        sync_cuda()
        epoch_train_time = time.perf_counter() - start

        sync_cuda()
        start = time.perf_counter()
        val_rmse, val_r2, true, predicted = evaluate(model, val_loader)
        sync_cuda()
        epoch_val_time = time.perf_counter() - start
        train_seconds += epoch_train_time
        val_seconds += epoch_val_time
        peak_cpu_mb = max(peak_cpu_mb, process.memory_info().rss / (1024**2))
        gpu_peak_mb = (torch.cuda.max_memory_allocated()/1024**2
                       if DEVICE.type == "cuda" else np.nan)

        history_rows.append({"strategy":strategy, "epoch":epoch,
                             "train_rmse":train_rmse, "train_r2":train_r2,
                             "val_rmse":val_rmse, "val_r2":val_r2,
                             "train_s":epoch_train_time, "validation_s":epoch_val_time,
                             "gpu_peak_mb":gpu_peak_mb})
        if val_rmse < best_rmse:
            best_rmse = val_rmse
            best_epoch = epoch
            best_predictions = (true.copy(), predicted.copy())
        print(f"Epoch {epoch:02d} | train RMSE {train_rmse:.3f} | val RMSE {val_rmse:.3f} | val R² {val_r2:.3f} | train {epoch_train_time:.2f}s")

    prediction_results[strategy] = best_predictions
    summary_rows.append({"strategy":strategy,
                         "best_val_rmse":best_rmse, "best_epoch":best_epoch,
                         "best_val_r2": next(row["val_r2"] for row in history_rows
                                             if row["strategy"]==strategy and row["epoch"]==best_epoch),
                         "train_s":train_seconds, "validation_s":val_seconds,
                         "total_s":train_seconds+val_seconds,
                         "gpu_peak_mb":gpu_peak_mb, "cpu_peak_mb":peak_cpu_mb,
                         "trainable_params":trainable, "total_params":total})
    del optimizer, model, train_loader, val_loader
    gc.collect()
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()

history = pd.DataFrame(history_rows)
summary = pd.DataFrame(summary_rows)
print("\nComparison summary (best validation RMSE and its corresponding R²):")
display(summary.round(3))
```

## 7. Validation metrics versus epoch

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for strategy in STRATEGIES:
    rows = history[history["strategy"] == strategy]
    axes[0].plot(rows["epoch"], rows["val_rmse"], marker="o", label=strategy)
    axes[1].plot(rows["epoch"], rows["val_r2"], marker="o", label=strategy)
axes[0].set(ylabel="Validation RMSE (log mol/L)", xlabel="Epoch", title="Validation RMSE vs epoch")
axes[1].set(ylabel="Validation R²", xlabel="Epoch", title="Validation R² vs epoch")
for ax in axes:
    ax.grid(alpha=0.25)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

## 8. Training time and memory comparison

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
names = summary["strategy"]
axes[0].bar(names, summary["train_s"], label="Training")
axes[0].bar(names, summary["validation_s"], bottom=summary["train_s"], label="Validation")
axes[0].set(ylabel="Total wall time (s)", title=f"Time for {EPOCHS} epochs")
axes[0].legend()
if DEVICE.type == "cuda":
    axes[1].bar(names, summary["gpu_peak_mb"])
    axes[1].set(ylabel="Peak GPU allocated memory (MiB)", title="Peak GPU memory")
else:
    axes[1].bar(names, summary["cpu_peak_mb"])
    axes[1].set(ylabel="Peak process RSS (MiB)", title="CPU memory (CUDA unavailable)")
for ax in axes:
    ax.tick_params(axis="x", labelrotation=25)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
for strategy in STRATEGIES:
    rows = history[history["strategy"]==strategy]
    plt.plot(rows["epoch"], rows["train_s"], marker=".", label=strategy)
plt.xlabel("Epoch"); plt.ylabel("Training time (s)")
plt.title("Training time per epoch"); plt.legend(fontsize=8)
plt.grid(alpha=0.25); plt.tight_layout(); plt.show()
```

## 9. Predicted vs. experimental solubility
Each panel uses the model predictions from the epoch with the lowest validation RMSE. As these epochs were selected on the validation set, report a separate test-set metric for an unbiased final generalization estimate.

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 10))
for ax, strategy in zip(axes.ravel(), STRATEGIES):
    true, predicted = prediction_results[strategy]
    lo = min(true.min(), predicted.min())
    hi = max(true.max(), predicted.max())
    ax.scatter(true, predicted, alpha=0.6, s=18)
    ax.plot([lo,hi], [lo,hi], "--", color="gray")
    ax.set(title=f"{strategy}\nRMSE={np.sqrt(mean_squared_error(true,predicted)):.3f}",
           xlabel="Measured log(solubility:mol/L)", ylabel="Predicted log(solubility:mol/L)")
    ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

## Notes on interpretation

- **Same data, same epochs, same batch size:** these make training costs easier to compare, though learning rates are deliberately tuned by strategy and may require further optimization for a rigorous performance benchmark.
- **Frozen backbone** still performs ChemBERTa inference each batch. Caching embeddings would make later epochs far faster but changes the cost definition; no caching is used here.
- **LoRA** keeps base weights frozen but gradients still pass through the encoder to compute LoRA updates; it primarily reduces trainable parameters and optimizer memory, not necessarily wall time.
- **No AMP:** all training uses the installed model's default FP32 precision. `torch.no_grad()` is used for validation only.
- **Partial freezing:** with three blocks, `floor(n_layers/2)` freezes one, not a literal 50% of blocks. For an exactly half-and-half example, choose a compatible checkpoint with an even number of encoder blocks.
- **Validation is not a test set.** For research reporting, keep a separate unseen test set, repeat multiple seeds, and consider scaffold splits for assessing generalization to new chemotypes.
